# Supermarket Sales Analytics
End-to-end exploratory data analysis using Python, Pandas and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/supermarket_sales.csv', parse_dates=['Date'])
df.head()
```

```python
print('Shape:', df.shape)
display(df.describe(include='all'))
print('\nMissing values:')
display(df.isna().sum())
```

```python
monthly = df.groupby(df['Date'].dt.to_period('M'))['Sales'].sum()
monthly.plot(kind='line', marker='o', figsize=(9,5), title='Monthly Sales Trend')
plt.ylabel('Sales (₹)'); plt.xlabel('Month'); plt.xticks(rotation=45); plt.show()
```

```python
category = df.groupby('Category').agg(Sales=('Sales','sum'), Profit=('Profit','sum')).sort_values('Sales', ascending=False)
display(category)
category['Sales'].plot(kind='bar', figsize=(9,5), title='Sales by Category')
plt.ylabel('Sales (₹)'); plt.xticks(rotation=25); plt.show()
```

```python
branch = df.groupby(['Branch','City']).agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Transactions=('Invoice_ID','count')).sort_values('Sales', ascending=False)
display(branch)
```

```python
payment = df['Payment_Method'].value_counts()
payment.plot(kind='bar', figsize=(8,5), title='Transactions by Payment Method')
plt.ylabel('Transactions'); plt.xticks(rotation=20); plt.show()
```
